### 28.3 Scrivere un file (modalità 'w')

```python
with open("esempio.txt", "w", encoding="utf-8") as file:
    file.write("Prima riga\n")
    file.write("Seconda riga\n")
print("File scritto.")
```

### 28.2 Leggere un file (modalità 'r')

```python
with open("esempio.txt", "r", encoding="utf-8") as file:
    contenuto = file.read()
print(contenuto)
```

Leggere riga per riga:

```python
with open("esempio.txt", "r", encoding="utf-8") as file:
    for riga in file:
        print(riga.strip())
```

### Aggiungere senza cancellare (modalità 'a')

```python
with open("esempio.txt", "a", encoding="utf-8") as file:
    file.write("Terza riga aggiunta dopo\n")

with open("esempio.txt", "r", encoding="utf-8") as file:
    print(file.read())
```
