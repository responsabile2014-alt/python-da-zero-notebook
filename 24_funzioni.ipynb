### 24.4 Funzione senza parametri

```python
def saluta():
    print("Ciao! Benvenuto nel programma.")
    print("Inizia quando sei pronto.")

saluta()
saluta()
```

### 24.5 Funzione con parametri

```python
def calcola_area_rettangolo(base, altezza):
    area = base * altezza
    print(f"Area del rettangolo {base} x {altezza} = {area}")

calcola_area_rettangolo(5, 3)
calcola_area_rettangolo(10, 4)
```

### 24.6 return

```python
def calcola_area(base, altezza):
    return base * altezza

area_sala = calcola_area(8, 5)
area_cucina = calcola_area(4, 3)
print(f"Area totale: {area_sala + area_cucina} m²")
```

### 24.7 print vs return

```python
def doppio_con_return(numero):
    return numero * 2

risultato = doppio_con_return(5)
print("Risultato:", risultato)   # 10
```

### 24.8 Parametri con valore di default

```python
def calcola_prezzo_con_iva(prezzo, iva=0.22):
    return prezzo * (1 + iva)

print(calcola_prezzo_con_iva(100))
print(round(calcola_prezzo_con_iva(100, 0.10), 2))
print(calcola_prezzo_con_iva(100, iva=0.04))
```

### ESEMPIO PRATICO – mini calcolatore

```python
def somma(a, b):
    return a + b

def divisione(a, b):
    if b == 0:
        return "Errore: divisione per zero"
    return a / b

print(somma(10, 5))
print(divisione(10, 5))
print(divisione(10, 0))
```

### 24.10 Type hints (opzionali)

```python
def somma(a: int, b: int) -> int:
    return a + b

print(somma(3, 4))
```
