### Passo 1 - Codice 'dell'AI' con un errore
Questo codice doveva contare le parole ma si blocca. Eseguilo e leggi l'errore.

```python
# Codice suggerito dall'AI (con errore voluto)
parole = ['ciao', 'mondo', 'ciao', 'python', 'ciao']
conteggio = {}
for p in parole:
    conteggio[p] = conteggio[p] + 1   # <-- questo dà errore!
print(conteggio)
```

### Passo 2 - Capiamo l'errore
La prima volta che incontra una parola nuova, la chiave non esiste ancora: KeyError. Ecco la correzione.

```python
parole = ['ciao', 'mondo', 'ciao', 'python', 'ciao']
conteggio = {}
for p in parole:
    conteggio[p] = conteggio.get(p, 0) + 1   # .get parte da 0 se nuova
print(conteggio)
```

```python
# Metodo: 1) esegui  2) leggi il tipo di errore  3) correggi  4) riprova
# KeyError -> chiave inesistente -> usa .get()
# NameError -> variabile non creata -> definiscila prima
# TypeError -> tipi mescolati -> converti con int/float/str
print("Hai imparato a non fidarti ciecamente del codice dell'AI")
```
