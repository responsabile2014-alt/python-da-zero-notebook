````python
def sconto(totale):
    if totale > 200:
        perc = 15
    elif totale > 100:
        perc = 10
    elif totale > 50:
        perc = 5
    else:
        perc = 0
    return totale * (1 - perc / 100)

for t in [40, 80, 150, 300]:
    print(f"{t} euro -> {sconto(t):.2f} euro")
````

**Output atteso:**
```
40 euro -> 40.00 euro
80 euro -> 76.00 euro
150 euro -> 135.00 euro
300 euro -> 255.00 euro
```

**Perché funziona:** Controllando dalla soglia più alta, assegni la percentuale corretta. La funzione restituisce il totale già scontato.

**Errore comune:** Controllare le soglie dalla più bassa: un importo di 300 prenderebbe subito il 5% invece del 15%.

**Variante da provare:** Aggiungi una soglia del 20% sopra i 500 euro.

### Esercizio 147. Convertitore di valuta con tabella

````python
tassi = {"USD": 1.08, "GBP": 0.85, "JPY": 160.0}
def converti(euro, valuta):
    if valuta not in tassi:
        return "Valuta non disponibile"
    return euro * tassi[valuta]

for v in ["USD", "GBP", "JPY"]:
    print(f"100 euro = {converti(100, v):.2f} {v}")
````

**Perché funziona:** Il dizionario è una tabella di conversione. La funzione cerca il tasso e moltiplica, gestendo le valute assenti.

**Variante da provare:** Aggiungi la conversione inversa: da valuta estera a euro.
