### Preparazione dati
Questo progetto usa il dataset **clienti_demo.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset clienti_demo.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'id;nome;citta;spesa_totale;attivo',
    '1;Anna Bianchi;Milano;1250.50;si',
    '2;Marco Rossi;Roma;890.00;si',
    '3;Sara Verdi;Milano;2300.75;si',
    '4;Luca Neri;Torino;450.20;no',
    '5;Elena Gialli;Roma;1780.00;si',
    '6;Paolo Blu;Napoli;320.50;no',
    '7;Giulia Viola;Milano;990.00;si',
    '8;Davide Rosa;Torino;1560.30;si',
    '9;Chiara Grigi;Napoli;670.80;no',
    '10;Matteo Bruni;Roma;2100.00;si',
    '11;Laura Conti;Milano;540.00;si',
    '12;Stefano Mari;Torino;1320.90;si'
]
with open('clienti_demo.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset clienti_demo.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Caricamento robusto con gestione errori
Leggiamo i clienti gestendo eventuali righe rotte.

```python
def carica_clienti(nomefile):
    clienti = []
    with open(nomefile, encoding='utf-8') as f:
        f.readline()
        for riga in f:
            try:
                idc, nome, citta, spesa, attivo = riga.strip().split(';')
                clienti.append({'nome': nome, 'citta': citta,
                                'spesa': float(spesa), 'attivo': attivo})
            except (ValueError, IndexError):
                continue
    return clienti

clienti = carica_clienti('clienti_demo.csv')
print(f'{len(clienti)} clienti caricati correttamente')
```

### Passo 2 - Analisi: clienti attivi e spesa
Quanti sono attivi e quanto spendono in media.

```python
attivi = [c for c in clienti if c['attivo'] == 'si']
spesa_media_attivi = sum(c['spesa'] for c in attivi) / len(attivi)
print(f'Clienti attivi: {len(attivi)} su {len(clienti)}')
print(f'Spesa media degli attivi: {spesa_media_attivi:.2f} euro')
```

### Passo 3 - Report finale per città
Spesa totale per città, ordinata dalla più alta.

```python
per_citta = {}
for c in clienti:
    per_citta[c['citta']] = per_citta.get(c['citta'], 0) + c['spesa']
ordinato = sorted(per_citta.items(), key=lambda x: x[1], reverse=True)
print('=== SPESA PER CITTÀ ===')
for citta, tot in ordinato:
    print(f'{citta:<10} {tot:>9.2f} euro')

top_cliente = max(clienti, key=lambda c: c['spesa'])
print(f'\nCliente top: {top_cliente["nome"]} ({top_cliente["spesa"]:.2f} euro)')
```
