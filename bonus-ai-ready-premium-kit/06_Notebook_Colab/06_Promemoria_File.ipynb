## Obiettivo
Salvare promemoria su un file di testo che resta anche dopo la chiusura.

## Prerequisiti
Aver studiato la scrittura di file e le liste.

## Concetti usati
scrittura file, append, persistenza dei dati

---

### Passo 1 - Salviamo i promemoria
Scriviamo alcuni promemoria su un file di testo.

```python
promemoria = ['Comprare il pane', 'Chiamare il dentista', 'Studiare Python']
with open('promemoria.txt', 'w', encoding='utf-8') as f:
    for p in promemoria:
        f.write(p + '\n')
print('Promemoria salvati su file')
```

### Passo 2 - Aggiungiamo un promemoria
Usiamo la modalità append per aggiungere senza cancellare.

```python
with open('promemoria.txt', 'a', encoding='utf-8') as f:
    f.write('Pagare la bolletta\n')
print('Promemoria aggiunto')
```

### Passo 3 - Leggiamo tutti i promemoria
Rileggiamo il file e li mostriamo numerati.

```python
with open('promemoria.txt', encoding='utf-8') as f:
    for i, riga in enumerate(f, 1):
        print(f'{i}. {riga.strip()}')
```
