### Preparazione dati
Questo progetto usa il dataset **prompt_log.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset prompt_log.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'data;argomento;prompt_usato;utile',
    '2025-01-10;Cicli;Spiegami il ciclo for con esempio semplice;si',
    '2025-01-11;Errori;Cosa significa IndexError;si',
    '2025-01-12;Liste;Differenza tra append e insert;si',
    '2025-01-13;Funzioni;Come restituire più valori da una funzione;si',
    '2025-01-14;Dizionari;Quando usare get invece delle parentesi;si',
    '2025-01-15;Debug;Trova l errore in questo codice;no',
    '2025-01-16;Stringhe;Come contare le vocali in una parola;si',
    '2025-01-17;File;Come leggere un file riga per riga;si',
    '2025-01-18;Condizioni;Differenza tra and e or;si',
    '2025-01-19;CSV;Come separare i campi di una riga CSV;si',
    '2025-01-20;Analisi;Come calcolare una media da una lista;si',
    "2025-01-21;AI;Come verificare se il codice dell'AI è corretto;si"
]
with open('prompt_log.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset prompt_log.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Carichiamo il log dei prompt
Ogni riga registra data, argomento, prompt e se è stato utile.

```python
log = []
with open('prompt_log.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        data, arg, prompt, utile = riga.strip().split(';')
        log.append({'argomento': arg, 'utile': utile})
print(f'{len(log)} prompt registrati')
```

### Passo 2 - Quanti prompt sono stati utili
Calcoliamo la percentuale di prompt utili.

```python
utili = sum(1 for p in log if p['utile'] == 'si')
perc = utili / len(log) * 100
print(f'Prompt utili: {utili}/{len(log)} ({perc:.0f}%)')
```

### Passo 3 - Argomenti su cui chiedo più aiuto
Raggruppiamo i prompt per argomento.

```python
per_argomento = {}
for p in log:
    per_argomento[p['argomento']] = per_argomento.get(p['argomento'], 0) + 1
top = max(per_argomento, key=lambda a: per_argomento[a])
for arg, n in per_argomento.items():
    print(f'{arg:<12} {n}')
print(f'\nArgomento su cui chiedo più aiuto: {top}')
```
