### Preparazione dati
Questo progetto usa il dataset **attivita_settimanali.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset attivita_settimanali.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'giorno;attivita;ore;completata',
    'Lunedì;Studio Python;2.0;si',
    'Lunedì;Palestra;1.0;si',
    'Martedì;Studio Python;1.5;si',
    'Martedì;Lettura;0.5;no',
    'Mercoledì;Studio Python;2.0;si',
    'Mercoledì;Progetto personale;1.0;no',
    'Giovedì;Studio Python;1.5;si',
    'Venerdì;Ripasso esercizi;2.0;si',
    'Venerdì;Palestra;1.0;si',
    'Sabato;Mini-progetto;3.0;si',
    'Domenica;Riposo;0.0;si',
    'Domenica;Pianificazione settimana;0.5;si'
]
with open('attivita_settimanali.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset attivita_settimanali.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Carichiamo le attività
Ogni riga ha giorno, attività, ore e se completata.

```python
attivita = []
with open('attivita_settimanali.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        giorno, att, ore, fatto = riga.strip().split(';')
        attivita.append({'giorno': giorno, 'ore': float(ore), 'completata': fatto})
print(f'{len(attivita)} attività caricate')
```

### Passo 2 - Ore totali e completamento
Quante ore in tutto e quante attività completate.

```python
ore_totali = sum(a['ore'] for a in attivita)
completate = sum(1 for a in attivita if a['completata'] == 'si')
perc = completate / len(attivita) * 100
print(f'Ore totali: {ore_totali:.1f}')
print(f'Completate: {completate}/{len(attivita)} ({perc:.0f}%)')
```

### Passo 3 - Ore per giorno
Raggruppiamo le ore per giorno della settimana.

```python
per_giorno = {}
for a in attivita:
    per_giorno[a['giorno']] = per_giorno.get(a['giorno'], 0) + a['ore']
top = max(per_giorno, key=lambda g: per_giorno[g])
for giorno, ore in per_giorno.items():
    print(f'{giorno:<11} {ore:.1f} ore')
print(f'\nGiorno più impegnato: {top}')
```
