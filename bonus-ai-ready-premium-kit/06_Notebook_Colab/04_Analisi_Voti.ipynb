### Preparazione dati
Questo progetto usa il dataset **studenti_voti.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset studenti_voti.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'studente;materia;voto',
    'Anna;Matematica;28',
    'Anna;Italiano;30',
    'Anna;Storia;26',
    'Marco;Matematica;22',
    'Marco;Italiano;18',
    'Marco;Storia;24',
    'Sara;Matematica;30',
    'Sara;Italiano;27',
    'Sara;Storia;29',
    'Luca;Matematica;19',
    'Luca;Italiano;21',
    'Luca;Storia;23'
]
with open('studenti_voti.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset studenti_voti.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Carichiamo i voti
Ogni riga ha studente, materia e voto.

```python
voti = []
with open('studenti_voti.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        studente, materia, voto = riga.strip().split(';')
        voti.append({'studente': studente, 'materia': materia, 'voto': int(voto)})
print(f'Caricati {len(voti)} voti')
```

### Passo 2 - Media per studente
Accumuliamo somma e conteggio per ogni studente.

```python
somme = {}
conteggi = {}
for v in voti:
    s = v['studente']
    somme[s] = somme.get(s, 0) + v['voto']
    conteggi[s] = conteggi.get(s, 0) + 1
for studente in somme:
    media = somme[studente] / conteggi[studente]
    print(f'{studente}: media {media:.2f}')
```

### Passo 3 - Media per materia
Stessa tecnica, ma raggruppando per materia.

```python
somme_m = {}
conteggi_m = {}
for v in voti:
    m = v['materia']
    somme_m[m] = somme_m.get(m, 0) + v['voto']
    conteggi_m[m] = conteggi_m.get(m, 0) + 1
for materia in somme_m:
    print(f'{materia}: media {somme_m[materia]/conteggi_m[materia]:.2f}')
```
