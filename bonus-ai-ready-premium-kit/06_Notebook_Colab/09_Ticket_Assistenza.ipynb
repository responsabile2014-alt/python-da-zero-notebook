### Preparazione dati
Questo progetto usa il dataset **ticket_assistenza.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset ticket_assistenza.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'id_ticket;cliente;problema;priorita;stato',
    'T001;Anna Bianchi;Login non funziona;alta;aperto',
    'T002;Marco Rossi;Password dimenticata;media;chiuso',
    'T003;Sara Verdi;Errore pagamento;alta;aperto',
    'T004;Luca Neri;Domanda fatturazione;bassa;chiuso',
    'T005;Elena Gialli;App si blocca;alta;in lavorazione',
    'T006;Paolo Blu;Richiesta rimborso;media;aperto',
    'T007;Giulia Viola;Modifica anagrafica;bassa;chiuso',
    'T008;Davide Rosa;Email non ricevuta;media;in lavorazione',
    'T009;Chiara Grigi;Account bloccato;alta;aperto',
    'T010;Matteo Bruni;Domanda generica;bassa;chiuso',
    'T011;Laura Conti;Errore caricamento;media;aperto',
    'T012;Stefano Mari;Problema stampa;bassa;in lavorazione'
]
with open('ticket_assistenza.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset ticket_assistenza.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Carichiamo i ticket
Ogni ticket ha cliente, problema, priorità e stato.

```python
ticket = []
with open('ticket_assistenza.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        idt, cliente, problema, prio, stato = riga.strip().split(';')
        ticket.append({'cliente': cliente, 'priorita': prio, 'stato': stato})
print(f'{len(ticket)} ticket caricati')
```

### Passo 2 - Quanti ticket per stato
Raggruppiamo per stato.

```python
per_stato = {}
for t in ticket:
    per_stato[t['stato']] = per_stato.get(t['stato'], 0) + 1
for stato, n in per_stato.items():
    print(f'{stato:<16} {n}')
```

### Passo 3 - Ticket urgenti da gestire
Troviamo i ticket aperti con priorità alta.

```python
print('URGENTI (aperti + priorità alta):')
for t in ticket:
    if t['stato'] == 'aperto' and t['priorita'] == 'alta':
        print(f"  {t['cliente']}")
```
