### Preparazione dati
Questo progetto usa il dataset **prodotti_magazzino.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset prodotti_magazzino.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'codice;prodotto;categoria;quantita;prezzo_unitario',
    'P001;Penna blu;Cancelleria;150;0.80',
    'P002;Quaderno A4;Cancelleria;85;2.50',
    'P003;Zaino scuola;Accessori;12;25.00',
    'P004;Astuccio;Accessori;8;7.50',
    'P005;Gomma;Cancelleria;200;0.40',
    'P006;Righello 30cm;Cancelleria;45;1.20',
    'P007;Calcolatrice;Elettronica;6;12.90',
    'P008;Evidenziatore;Cancelleria;95;1.50',
    'P009;Cartella;Accessori;20;9.00',
    'P010;Pennarelli set;Cancelleria;30;6.80',
    'P011;Diario;Cancelleria;18;8.50',
    'P012;Forbici;Accessori;9;3.20'
]
with open('prodotti_magazzino.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset prodotti_magazzino.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Carichiamo il magazzino
Ogni prodotto ha codice, nome, categoria, quantità e prezzo.

```python
magazzino = []
with open('prodotti_magazzino.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        cod, prod, cat, qta, prezzo = riga.strip().split(';')
        magazzino.append({'prodotto': prod, 'categoria': cat, 'quantita': int(qta), 'prezzo': float(prezzo)})
print(f'{len(magazzino)} prodotti caricati')
```

### Passo 2 - Valore totale del magazzino
Il valore di ogni prodotto è quantità per prezzo.

```python
valore_totale = sum(p['quantita'] * p['prezzo'] for p in magazzino)
print(f'Valore totale magazzino: {valore_totale:.2f} euro')
```

### Passo 3 - Scorte basse
Segnaliamo i prodotti sotto le 20 unita'.

```python
print('Prodotti da riordinare (sotto 20 pezzi):')
for p in magazzino:
    if p['quantita'] < 20:
        print(f"  {p['prodotto']}: {p['quantita']} pezzi")
```
