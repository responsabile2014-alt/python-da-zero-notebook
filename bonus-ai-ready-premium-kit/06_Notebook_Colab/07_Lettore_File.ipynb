Leggere un file di testo e calcolare statistiche: righe, parole, caratteri.

---

### Passo 1 - Creiamo un file di esempio
Scriviamo un breve testo da analizzare.

```python
testo = '''Python e un linguaggio potente.
E anche semplice da imparare.
Migliaia di persone lo usano ogni giorno.'''
with open('articolo.txt', 'w', encoding='utf-8') as f:
    f.write(testo)
print('File creato')
```

### Passo 2 - Contiamo righe, parole e caratteri
Leggiamo il file e calcoliamo le statistiche.

```python
righe = 0
parole = 0
caratteri = 0
with open('articolo.txt', encoding='utf-8') as f:
    for riga in f:
        righe += 1
        parole += len(riga.split())
        caratteri += len(riga.strip())
print(f'Righe: {righe}')
print(f'Parole: {parole}')
print(f'Caratteri: {caratteri}')
```

### Passo 3 - La parola più lunga
Troviamo la parola più lunga del testo.

```python
piu_lunga = ''
with open('articolo.txt', encoding='utf-8') as f:
    for riga in f:
        for parola in riga.split():
            if len(parola) > len(piu_lunga):
                piu_lunga = parola
print(f'Parola più lunga: {piu_lunga} ({len(piu_lunga)} caratteri)')
```
