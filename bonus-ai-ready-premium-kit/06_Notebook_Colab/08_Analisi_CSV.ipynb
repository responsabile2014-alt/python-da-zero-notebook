### Preparazione dati
Questo progetto usa il dataset **vendite_mensili.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset vendite_mensili.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'mese;prodotto;quantita_venduta;incasso',
    'Gennaio;Penna blu;320;256.00',
    'Gennaio;Quaderno A4;145;362.50',
    'Gennaio;Zaino scuola;18;450.00',
    'Febbraio;Penna blu;280;224.00',
    'Febbraio;Quaderno A4;160;400.00',
    'Febbraio;Zaino scuola;22;550.00',
    'Marzo;Penna blu;410;328.00',
    'Marzo;Quaderno A4;190;475.00',
    'Marzo;Zaino scuola;15;375.00',
    'Aprile;Penna blu;350;280.00',
    'Aprile;Quaderno A4;210;525.00',
    'Aprile;Zaino scuola;28;700.00'
]
with open('vendite_mensili.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset vendite_mensili.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Leggiamo il CSV
Convertiamo i campi numerici (qui sta l'errore più comune).

```python
vendite = []
with open('vendite_mensili.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        mese, prod, qta, incasso = riga.strip().split(';')
        vendite.append({'mese': mese, 'prodotto': prod, 'quantita': int(qta), 'incasso': float(incasso)})
print(f'{len(vendite)} righe di vendita caricate')
```

### Passo 2 - Incasso totale per mese
Raggruppiamo gli incassi per mese.

```python
per_mese = {}
for v in vendite:
    per_mese[v['mese']] = per_mese.get(v['mese'], 0) + v['incasso']
for mese, tot in per_mese.items():
    print(f'{mese:<10} {tot:>8.2f} euro')
```

### Passo 3 - Prodotto più venduto in assoluto
Sommiamo le quantità per prodotto e troviamo il migliore.

```python
per_prodotto = {}
for v in vendite:
    per_prodotto[v['prodotto']] = per_prodotto.get(v['prodotto'], 0) + v['quantita']
top = max(per_prodotto, key=lambda p: per_prodotto[p])
print(f'Prodotto più venduto: {top} ({per_prodotto[top]} pezzi)')
```
