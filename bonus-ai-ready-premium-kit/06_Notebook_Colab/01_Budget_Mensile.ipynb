### Preparazione dati
Questo progetto usa il dataset **budget_familiare.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset budget_familiare.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'voce;tipo;importo_previsto;importo_reale',
    'Stipendio;entrata;2200.00;2200.00',
    'Affitto;uscita;750.00;750.00',
    'Bollette;uscita;180.00;210.50',
    'Spesa alimentare;uscita;450.00;478.30',
    'Trasporti;uscita;120.00;135.00',
    'Svago;uscita;150.00;95.00',
    'Risparmio;uscita;300.00;300.00',
    'Rimborso spese;entrata;80.00;80.00',
    'Assicurazione;uscita;90.00;90.00',
    'Telefono e internet;uscita;45.00;45.00',
    'Palestra;uscita;40.00;40.00',
    'Extra;uscita;100.00;165.20'
]
with open('budget_familiare.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset budget_familiare.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - I dati del budget
Leggiamo il budget da un file CSV. Ogni voce ha tipo (entrata/uscita) e importo.

```python
voci = []
with open('budget_familiare.csv', encoding='utf-8') as f:
    f.readline()  # salto intestazione
    for riga in f:
        nome, tipo, previsto, reale = riga.strip().split(';')
        voci.append({'nome': nome, 'tipo': tipo, 'reale': float(reale)})
print(f'Caricate {len(voci)} voci')
```

### Passo 2 - Calcoliamo entrate, uscite e saldo
Sommiamo separatamente entrate e uscite usando una condizione.

```python
entrate = sum(v['reale'] for v in voci if v['tipo'] == 'entrata')
uscite = sum(v['reale'] for v in voci if v['tipo'] == 'uscita')
saldo = entrate - uscite
print(f'Entrate: {entrate:.2f} euro')
print(f'Uscite: {uscite:.2f} euro')
print(f'Saldo: {saldo:.2f} euro')
```

### Passo 3 - Un report ordinato
Stampiamo ogni uscita con la sua percentuale sul totale delle uscite.

```python
print('=== DETTAGLIO USCITE ===')
for v in voci:
    if v['tipo'] == 'uscita':
        perc = v['reale'] / uscite * 100
        print(f"{v['nome']:<20} {v['reale']:>8.2f} euro ({perc:4.1f}%)")
```
