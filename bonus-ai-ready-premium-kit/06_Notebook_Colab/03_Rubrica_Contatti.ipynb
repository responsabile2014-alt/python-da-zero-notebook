### Preparazione dati
Questo progetto usa il dataset **clienti_demo.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset clienti_demo.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'id;nome;citta;spesa_totale;attivo',
    '1;Anna Bianchi;Milano;1250.50;si',
    '2;Marco Rossi;Roma;890.00;si',
    '3;Sara Verdi;Milano;2300.75;si',
    '4;Luca Neri;Torino;450.20;no',
    '5;Elena Gialli;Roma;1780.00;si',
    '6;Paolo Blu;Napoli;320.50;no',
    '7;Giulia Viola;Milano;990.00;si',
    '8;Davide Rosa;Torino;1560.30;si',
    '9;Chiara Grigi;Napoli;670.80;no',
    '10;Matteo Bruni;Roma;2100.00;si',
    '11;Laura Conti;Milano;540.00;si',
    '12;Stefano Mari;Torino;1320.90;si'
]
with open('clienti_demo.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset clienti_demo.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Carichiamo i contatti
Usiamo il file clienti come rubrica.

```python
rubrica = []
with open('clienti_demo.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        idc, nome, citta, spesa, attivo = riga.strip().split(';')
        rubrica.append({'nome': nome, 'citta': citta})
print(f'{len(rubrica)} contatti in rubrica')
```

### Passo 2 - Una funzione di ricerca
Scriviamo una funzione che cerca un contatto per nome.

```python
def cerca(rubrica, testo):
    risultati = []
    for c in rubrica:
        if testo.lower() in c['nome'].lower():
            risultati.append(c)
    return risultati

trovati = cerca(rubrica, 'anna')
for c in trovati:
    print(f"{c['nome']} - {c['citta']}")
```

### Passo 3 - Contatti per città
Contiamo quanti contatti ci sono per ogni città.

```python
per_citta = {}
for c in rubrica:
    per_citta[c['citta']] = per_citta.get(c['citta'], 0) + 1
for citta, n in per_citta.items():
    print(f'{citta}: {n} contatti')
```
