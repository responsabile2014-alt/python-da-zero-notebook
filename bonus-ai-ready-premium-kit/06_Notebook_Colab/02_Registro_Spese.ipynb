### Preparazione dati
Questo progetto usa il dataset **spese_personali.csv**. La cella qui sotto lo crea automaticamente, così il notebook funziona subito in Colab.

```python
# Prepariamo il dataset spese_personali.csv (in Colab questa cella crea il file da usare)
righe_csv = [
    'data;categoria;descrizione;importo',
    '2025-01-03;Alimentari;Spesa supermercato;52.30',
    '2025-01-05;Trasporti;Abbonamento bus;35.00',
    '2025-01-08;Svago;Cinema;9.50',
    '2025-01-12;Alimentari;Pane e latte;6.80',
    '2025-01-15;Casa;Detersivi;14.20',
    '2025-01-18;Salute;Farmacia;23.00',
    '2025-01-21;Svago;Libro;16.90',
    '2025-01-24;Alimentari;Spesa settimanale;61.40',
    '2025-01-27;Trasporti;Benzina;45.00',
    '2025-01-30;Ristorante;Cena fuori;38.50',
    '2025-02-02;Alimentari;Frutta e verdura;18.70',
    '2025-02-05;Casa;Lampadine;8.90'
]
with open('spese_personali.csv', 'w', encoding='utf-8') as f:
    f.write('\n'.join(righe_csv))
print('Dataset spese_personali.csv pronto:', len(righe_csv), 'righe')
```

### Passo 1 - Carichiamo le spese
Ogni spesa ha data, categoria, descrizione e importo.

```python
spese = []
with open('spese_personali.csv', encoding='utf-8') as f:
    f.readline()
    for riga in f:
        data, cat, descr, imp = riga.strip().split(';')
        spese.append({'categoria': cat, 'importo': float(imp)})
print(f'Caricate {len(spese)} spese')
```

### Passo 2 - Totale per categoria
Usiamo un dizionario come 'registro' che accumula la spesa per ogni categoria.

```python
per_categoria = {}
for s in spese:
    cat = s['categoria']
    per_categoria[cat] = per_categoria.get(cat, 0) + s['importo']
for cat, tot in per_categoria.items():
    print(f'{cat:<14} {tot:>7.2f} euro')
```

### Passo 3 - La categoria più costosa
Troviamo dove spendiamo di più.

```python
totale = sum(per_categoria.values())
top = max(per_categoria, key=lambda c: per_categoria[c])
print(f'Spesa totale: {totale:.2f} euro')
print(f'Categoria più costosa: {top} ({per_categoria[top]:.2f} euro)')
```
