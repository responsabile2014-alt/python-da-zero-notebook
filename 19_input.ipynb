### 19.2 input() di base

### 19.3–19.4 input() è SEMPRE una stringa: convertila con int() o float()

```python
eta = int(input("Quanti anni hai? "))
print("L'anno prossimo avrai", eta + 1, "anni")
```

```python
prezzo = float(input("Inserisci il prezzo del prodotto: "))
print(f"Prezzo con IVA: {prezzo * 1.22:.2f} €")
```

### 19.4 Virgola italiana: sostituiscila con il punto prima di float()

```python
prezzo_testo = input("Prezzo (es. 12,50): ")
prezzo = float(prezzo_testo.replace(",", "."))
print(f"Prezzo inserito: {prezzo:.2f} €")
```

### ESEMPIO PRATICO – calcolatore di età

```python
from datetime import date
nome = input("Come ti chiami? ")
anno_nascita = int(input("In che anno sei nato/a? "))
eta = date.today().year - anno_nascita
print(f"Ciao {nome}! Hai circa {eta} anni.")
```
