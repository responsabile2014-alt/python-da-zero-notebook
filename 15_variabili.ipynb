### 15.5 Leggere il valore di una variabile (senza virgolette!)

```python
nome = "Anna"
print(nome)    # Anna  (il valore)
print("nome")  # nome  (la parola come testo)
```

### 15.6 Variabile + testo: virgola oppure f-string

```python
nome = "Anna"
eta = 30
print("Il mio nome è", nome)
print(f"Ho {eta} anni")
```

### 15.7 Aggiornare una variabile

```python
contatore = 0
print("Contatore:", contatore)
contatore = contatore + 1
print("Contatore dopo:", contatore)
```

### 15.8 Nomi delle variabili
Regola: lettere, numeri e underscore; non iniziare con un numero; niente spazi; **niente accenti**. Esempi:

| Corretto | Sbagliato |
|---|---|
| `eta = 30` | `età = 30` |
| `numero_studenti = 12` | `numero studenti = 12` |
| `numero_1 = 10` | `1numero = 10` |

### 15.12 Assegnazione multipla

```python
nome, eta, citta = "Anna", 30, "Roma"
print(nome)
print(eta)
print(citta)
```
