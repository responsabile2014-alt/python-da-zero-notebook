Gli errori non sono nemici: ti dicono **cosa**, **dove** e **perché**. Qui sotto i 8 errori più comuni, mostrati come testo (codice sbagliato → messaggio → correzione). Provali pure tu in una cella nuova per vedere il messaggio dal vivo.

**SyntaxError** – grammatica sbagliata
```python
print("Ciao"        # SyntaxError: '(' was never closed
if eta >= 18         # SyntaxError: expected ':'
nome = "Anna          # SyntaxError: unterminated string literal
```

**NameError** – nome non definito / maiuscola sbagliata
```python
print(nome)   # NameError: name 'nome' is not defined
Print("Ciao") # NameError: name 'Print' is not defined
```

**TypeError** – tipi incompatibili
```python
"Ho " + 30          # TypeError: can only concatenate str (not "int") to str
for i in 5: ...      # TypeError: 'int' object is not iterable
```

**ValueError** – valore non convertibile
```python
int("trenta")   # ValueError: invalid literal for int() with base 10: 'trenta'
float("19,99")  # ValueError: could not convert string to float: '19,99'
```

**IndentationError** – spazi sbagliati • **IndexError** – indice fuori range • **KeyError** – chiave assente • **AttributeError** – metodo inesistente
```python
if True:
print("Ciao")             # IndentationError: expected an indented block
["a","b"][5]               # IndexError: list index out of range
{"nome":"Anna"}["email"]   # KeyError: 'email'
(42).upper()               # AttributeError: 'int' object has no attribute 'upper'
```
