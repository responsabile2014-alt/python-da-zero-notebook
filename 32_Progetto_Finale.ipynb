## Cella 1 – Configurazione iniziale

```python
# ============================================
# GESTIONE SPESE PERSONALI - Mini Progetto Finale
# ============================================
# --- CONFIGURAZIONE ---
BUDGET_MENSILE = 1000.00
CATEGORIE_VALIDE = [
    "alimentari", "trasporti", "salute", "svago",
    "abbigliamento", "casa", "altro"
]
spese = []  # qui memorizziamo tutte le spese (lista di dizionari)

print("Configurazione caricata.")
print(f"Budget mensile: {BUDGET_MENSILE:.2f} €")
print(f"Categorie disponibili: {', '.join(CATEGORIE_VALIDE)}")
```

## Cella 2 – Funzioni di supporto

```python
# --- FUNZIONI DI SUPPORTO ---

def mostra_menu():
    """Mostra il menu principale del programma."""
    print("\n" + "=" * 45)
    print("   GESTIONE SPESE PERSONALI")
    print("=" * 45)
    print("1 - Aggiungi una spesa")
    print("2 - Visualizza tutte le spese")
    print("3 - Riepilogo per categoria")
    print("4 - Analisi statistiche")
    print("5 - Controllo budget")
    print("6 - Cancella tutte le spese")
    print("0 - Esci")
    print("=" * 45)

def aggiungi_spesa():
    """Chiede i dettagli di una spesa e la aggiunge alla lista. Restituisce True/False."""
    print("\n--- NUOVA SPESA ---")
    descrizione = input("Descrizione della spesa: ").strip()
    if not descrizione:
        print("Errore: la descrizione non può essere vuota.")
        return False
    print("\nCategorie disponibili:")
    for i, cat in enumerate(CATEGORIE_VALIDE, 1):
        print(f"  {i}. {cat}")
    scelta_cat = input("Categoria (numero o nome): ").strip().lower()
    if scelta_cat.isdigit():
        indice = int(scelta_cat) - 1
        if 0 <= indice < len(CATEGORIE_VALIDE):
            categoria = CATEGORIE_VALIDE[indice]
        else:
            print("Errore: numero categoria non valido.")
            return False
    else:
        categoria = scelta_cat
    if categoria not in CATEGORIE_VALIDE:
        print(f"Errore: categoria '{categoria}' non valida.")
        print(f"Usa una di: {', '.join(CATEGORIE_VALIDE)}")
        return False
    try:
        importo_testo = input("Importo (€): ").strip().replace(",", ".")
        importo = float(importo_testo)
    except ValueError:
        print("Errore: l'importo deve essere un numero. Esempio: 12.50 oppure 12,50")
        return False
    if importo <= 0:
        print("Errore: l'importo deve essere maggiore di zero.")
        return False
    spesa = {"descrizione": descrizione, "categoria": categoria, "importo": importo}
    spese.append(spesa)
    print(f"\nOK Spesa aggiunta: '{descrizione}'  {importo:.2f} € ({categoria})")
    return True

def mostra_tutte_le_spese():
    """Mostra tutte le spese inserite."""
    print("\n--- ELENCO SPESE ---")
    if len(spese) == 0:
        print("Nessuna spesa inserita.")
        return
    header = f"{'N':>3} {'Descrizione':<25} {'Categoria':<15} {'Importo':>10}"
    print("\n" + header)
    print("-" * 57)
    for i, spesa in enumerate(spese, 1):
        print(f"{i:>3} {spesa['descrizione']:<25} {spesa['categoria']:<15} {spesa['importo']:>9.2f}€")
    totale = sum(s["importo"] for s in spese)
    print("-" * 57)
    print(f"{'TOTALE':>47} {totale:>9.2f}€")

def riepilogo_per_categoria():
    """Calcola e mostra il totale speso per ogni categoria."""
    print("\n--- RIEPILOGO PER CATEGORIA ---")
    if len(spese) == 0:
        print("Nessuna spesa inserita.")
        return
    totali_categorie = {}
    for spesa in spese:
        cat = spesa["categoria"]
        totali_categorie[cat] = totali_categorie.get(cat, 0) + spesa["importo"]
    totale_generale = sum(totali_categorie.values())
    print(f"\n{'Categoria':<20} {'Totale':>10} {'%':>8}")
    print("-" * 42)
    voci_ordinate = sorted(totali_categorie.items(), key=lambda x: x[1], reverse=True)
    for cat, totale in voci_ordinate:
        percentuale = (totale / totale_generale) * 100
        print(f"{cat:<20} {totale:>9.2f}€ {percentuale:>7.1f}%")
    print("-" * 42)
    print(f"{'TOTALE':<20} {totale_generale:>9.2f}€ {'100.0':>7}%")

def analizza_spese():
    """Calcola statistiche sulle spese: min, max, media."""
    print("\n--- ANALISI SPESE ---")
    if len(spese) == 0:
        print("Nessuna spesa inserita.")
        return
    importi = [s["importo"] for s in spese]
    spesa_massima = max(importi)
    spesa_minima = min(importi)
    totale = sum(importi)
    media = totale / len(importi)
    nome_max = nome_min = ""
    for s in spese:
        if s["importo"] == spesa_massima:
            nome_max = s["descrizione"]
        if s["importo"] == spesa_minima:
            nome_min = s["descrizione"]
    print(f"\nNumero spese: {len(spese)}")
    print(f"Spesa totale: {totale:.2f} €")
    print(f"Spesa media: {media:.2f} €")
    print(f"Spesa massima: {spesa_massima:.2f} € ({nome_max})")
    print(f"Spesa minima: {spesa_minima:.2f} € ({nome_min})")

def controlla_budget():
    """Confronta il totale spese con il budget mensile."""
    print("\n--- CONTROLLO BUDGET ---")
    totale_speso = sum(s["importo"] for s in spese)
    rimanente = BUDGET_MENSILE - totale_speso
    percentuale_usata = (totale_speso / BUDGET_MENSILE) * 100
    print(f"Budget mensile: {BUDGET_MENSILE:.2f} €")
    print(f"Totale speso:   {totale_speso:.2f} €")
    print(f"Rimanente:      {rimanente:.2f} €")
    print(f"Budget usato:   {percentuale_usata:.1f}%")
    barre_piene = min(int(percentuale_usata / 5), 20)
    barra = "#" * barre_piene + "." * (20 - barre_piene)
    print(f"[{barra}] {percentuale_usata:.1f}%")
    if totale_speso > BUDGET_MENSILE:
        print(f"\n[!] Budget SUPERATO di {totale_speso - BUDGET_MENSILE:.2f} €!")
    elif percentuale_usata >= 80:
        print("\n[!] Hai usato l'80% del budget!")
    elif percentuale_usata >= 50:
        print("\n-> Sei a metà budget.")
    else:
        print("\nOK Budget sotto controllo.")

print("Funzioni definite correttamente.")
```

## Cella 3 – Programma principale (interattivo)
> Suggerimento: prova ad aggiungere 2–3 spese (opzione 1), poi guarda *Riepilogo* (3), *Analisi* (4) e *Budget* (5).

```python
# --- PROGRAMMA PRINCIPALE ---
print("Programma Gestione Spese Personali avviato!")
print(f"Budget mensile impostato: {BUDGET_MENSILE:.2f} €")

while True:
    mostra_menu()
    scelta = input("\nScegli un'opzione (0-6): ").strip()
    if scelta == "1":
        aggiungi_spesa()
    elif scelta == "2":
        mostra_tutte_le_spese()
    elif scelta == "3":
        riepilogo_per_categoria()
    elif scelta == "4":
        analizza_spese()
    elif scelta == "5":
        controlla_budget()
    elif scelta == "6":
        if len(spese) == 0:
            print("La lista è già vuota.")
        else:
            conferma = input("Confermi? (si/no): ").strip().lower()
            if conferma == "si":
                spese.clear()
                print("Tutte le spese sono state cancellate.")
            else:
                print("Operazione annullata.")
    elif scelta == "0":
        print("\nArrivederci! Il programma è terminato.")
        print(f"Spese inserite in questa sessione: {len(spese)}")
        if len(spese) > 0:
            print(f"Totale speso: {sum(s['importo'] for s in spese):.2f} €")
        break
    else:
        print("Opzione non valida (0-6).")
```

## (Opzionale) Estensione 32.9 – Salvataggio in CSV

```python
# --- ESTENSIONE OPZIONALE: salvare le spese in CSV (sez. 32.9) ---
import csv
from pathlib import Path

NOME_FILE = "spese.csv"

def salva_spesa_csv(spesa):
    file_esiste = Path(NOME_FILE).exists()
    with open(NOME_FILE, "a", encoding="utf-8", newline="") as file:
        writer = csv.writer(file, delimiter=";")
        if not file_esiste:
            writer.writerow(["descrizione", "categoria", "importo"])
        writer.writerow([spesa["descrizione"], spesa["categoria"], f"{spesa['importo']:.2f}"])

# Per usarla: dentro aggiungi_spesa(), dopo spese.append(spesa), aggiungi:
#     salva_spesa_csv(spesa)
print("Funzione salva_spesa_csv pronta.")
```
