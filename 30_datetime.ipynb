### 30.2–30.3 Data di oggi; anno, mese, giorno

```python
from datetime import date
oggi = date.today()
print(oggi)
print("Anno:", oggi.year)
print("Mese:", oggi.month)
print("Giorno:", oggi.day)
```

### 30.4 Creare una data (ordine: anno, mese, giorno)

```python
from datetime import date
data_nascita = date(1990, 3, 15)
print(data_nascita)
```

### 30.5 Età esatta (confronto mese/giorno con le tuple)

```python
from datetime import date
nascita = date(1994, 3, 15)
oggi = date.today()
eta = oggi.year - nascita.year
if (oggi.month, oggi.day) < (nascita.month, nascita.day):
    eta = eta - 1
print(f"Età esatta: {eta} anni")
```

### 30.6 Formato italiano con strftime()

```python
from datetime import date
print(date.today().strftime("%d/%m/%Y"))
```

### 30.7 Registrare una spesa con data automatica (interattivo)

```python
from datetime import date
descrizione = input("Descrizione spesa: ")
importo = float(input("Importo: ").replace(",", "."))
data_spesa = date.today().strftime("%d/%m/%Y")
print(f"{data_spesa} - {descrizione}: {importo:.2f} euro")
```
