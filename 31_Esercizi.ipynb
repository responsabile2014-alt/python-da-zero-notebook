## LIVELLO 7 – Mini Progetti (interattivi)

**7.1** Calcolatore dell'età.

```python
from datetime import date
nome = input("Come ti chiami? ")
anno_nascita = int(input("Anno di nascita (es. 1994): "))
mese_nascita = int(input("Mese di nascita (1-12): "))
giorno_nascita = int(input("Giorno di nascita (1-31): "))
oggi = date.today()
eta = oggi.year - anno_nascita
if (oggi.month, oggi.day) < (mese_nascita, giorno_nascita):
    eta -= 1
print(f"\nCiao {nome}! Hai {eta} anni.")
print(f"Mancano {100 - eta} anni per arrivare a 100 anni!")
```

**7.2** Lista della spesa (menu con while).

```python
spesa = []
print("=== LISTA DELLA SPESA ===")
while True:
    print("\n1 - Aggiungi  2 - Rimuovi  3 - Visualizza  4 - Esci")
    scelta = input("Scelta (1-4): ").strip()
    if scelta == "1":
        prodotto = input("Prodotto da aggiungere: ").strip()
        if prodotto:
            spesa.append(prodotto)
            print(f"'{prodotto}' aggiunto.")
    elif scelta == "2":
        prodotto = input("Prodotto da rimuovere: ").strip()
        if prodotto in spesa:
            spesa.remove(prodotto)
            print(f"'{prodotto}' rimosso.")
        else:
            print("Non presente.")
    elif scelta == "3":
        if not spesa:
            print("La lista è vuota.")
        else:
            for i, p in enumerate(spesa, 1):
                print(f"  {i}. {p}")
    elif scelta == "4":
        print("Arrivederci!")
        break
    else:
        print("Scelta non valida.")
```

**7.3** Registro studenti.

```python
registro = {}
NUM_STUDENTI = 2
NUM_VOTI = 3
for i in range(NUM_STUDENTI):
    nome = input(f"\nNome dello studente {i+1}: ").strip()
    voti = []
    for j in range(NUM_VOTI):
        voti.append(int(input(f"  Voto {j+1} di {nome}: ")))
    registro[nome] = voti
for nome, voti in registro.items():
    media = sum(voti) / len(voti)
    print(f"\n{nome}: voti {voti}, media {media:.1f} -> {'PROMOSSO' if media >= 6 else 'BOCCIATO'}")
```

**7.6** Rubrica telefonica.

```python
rubrica = {"Anna Rossi": "333-111-2222", "Marco Verdi": "347-555-6666"}
print("=== RUBRICA TELEFONICA ===")
while True:
    print("\n1 - Aggiungi  2 - Cerca  3 - Visualizza  4 - Elimina  5 - Esci")
    scelta = input("Scelta: ").strip()
    if scelta == "1":
        nome = input("Nome contatto: ").strip()
        rubrica[nome] = input("Numero: ").strip()
        print("Aggiunto.")
    elif scelta == "2":
        nome = input("Nome da cercare: ").strip()
        numero = rubrica.get(nome)
        print(f"{nome}: {numero}" if numero else "Non trovato.")
    elif scelta == "3":
        for nome, numero in sorted(rubrica.items()):
            print(f"  {nome}: {numero}")
    elif scelta == "4":
        nome = input("Nome da eliminare: ").strip()
        if nome in rubrica:
            del rubrica[nome]
            print("Eliminato.")
        else:
            print("Non trovato.")
    elif scelta == "5":
        print("Arrivederci!")
        break
    else:
        print("Scelta non valida.")
```

**7.7** Analisi statistica (interattivo).

```python
numeri = []
print("Inserisci i numeri ('fine' per terminare)")
while True:
    v = input("Numero (o 'fine'): ").strip()
    if v.lower() == "fine":
        break
    numeri.append(float(v.replace(",", ".")))
if not numeri:
    print("Nessun numero inserito.")
else:
    numeri_ordinati = sorted(numeri)
    mid = len(numeri) // 2
    if len(numeri) % 2 == 0:
        mediana = (numeri_ordinati[mid - 1] + numeri_ordinati[mid]) / 2
    else:
        mediana = numeri_ordinati[mid]
    print(f"Conteggio: {len(numeri)}")
    print(f"Somma: {sum(numeri):.2f}")
    print(f"Media: {sum(numeri)/len(numeri):.2f}")
    print(f"Mediana: {mediana:.2f}")
    print(f"Minimo: {min(numeri):.2f}  Massimo: {max(numeri):.2f}")
```

**8.13** Riepilogo: data + input sicuro + salvataggio CSV (interattivo).

```python
from datetime import date
nome_file = "spese_semplici.csv"
data_oggi = date.today().strftime("%d/%m/%Y")
descrizione = input("Descrizione spesa: ")
importo_testo = input("Importo spesa: ")
try:
    importo = float(importo_testo.replace(",", "."))
    with open(nome_file, "a", encoding="utf-8") as file:
        file.write(f"{data_oggi};{descrizione};{importo:.2f}\n")
    print("Spesa salvata correttamente.")
    print(f"{data_oggi} - {descrizione}: {importo:.2f} euro")
except ValueError:
    print("Errore: l'importo deve essere un numero. Esempio: 12.50 oppure 12,50")
```
