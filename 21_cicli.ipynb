### 21.2 Ciclo for su una lista e su una stringa

````python
frutti = ["mela", "banana", "arancia"]
for frutto in frutti:
    print("Sto mangiando:", frutto)
````

````python
for lettera in "Python":
    print(lettera)
````

### 21.3 range()

````python
for i in range(5):
    print(i)
````

````python
for i in range(1, 6):
    print(i)
````

````python
for i in range(0, 10, 2):
    print(i)
````

````python
for i in range(5, 0, -1):
    print(i)
````

### 21.4 Ciclo while (ricordati di aggiornare la variabile!)

````python
contatore = 1
while contatore <= 5:
    print("Contatore:", contatore)
    contatore += 1
````

**Ciclo infinito (NON eseguire):** manca l'aggiornamento → gira per sempre.
```python

```

### 21.7 break

````python
for i in range(10):
    if i == 5:
        break
    print(i)
print("Ciclo terminato")
````

### 21.8 continue

````python
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
````

### 21.9 Cicli annidati

````python
for riga in range(1, 4):
    for colonna in range(1, 4):
        if colonna < 3:
            print(f"({riga},{colonna})", end=" ")
        else:
            print(f"({riga},{colonna})")
````

### ESEMPIO PRATICO – somma e media di una lista

````python
numeri = [10, 25, 7, 43, 18, 2]
totale = 0
for numero in numeri:
    totale += numero
print(f"Somma: {totale}")
print(f"Media: {totale / len(numeri):.2f}")
````
