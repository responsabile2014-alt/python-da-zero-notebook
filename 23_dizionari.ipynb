### 23.1–23.2 Creare e leggere per chiave

```python
persona = {"nome": "Anna", "cognome": "Rossi", "eta": 30, "citta": "Roma"}
print(persona)
print(persona["nome"])
print(persona["eta"])
```

### 23.3 KeyError e la soluzione con .get()

```python
persona = {"nome": "Anna", "eta": 30}
print(persona.get("email"))         # None, nessun errore
print(persona.get("email", "N/A"))  # valore di default
```

### 23.4–23.5 Aggiungere, modificare, rimuovere

```python
persona = {"nome": "Anna", "eta": 30}
persona["email"] = "anna@email.it"
print(persona)
persona["eta"] = 31
print(persona)
del persona["email"]
print(persona)
```

### 23.6 Iterare su un dizionario

```python
persona = {"nome": "Anna", "eta": 30, "citta": "Roma"}
for chiave, valore in persona.items():
    print(f"{chiave}: {valore}")
```

### 23.8 Lista di dizionari

```python
studenti = [
    {"nome": "Anna", "voto": 85},
    {"nome": "Marco", "voto": 72},
    {"nome": "Sofia", "voto": 91},
]
for studente in studenti:
    print(f"{studente['nome']}: {studente['voto']}")
```

### ESEMPIO PRATICO – mini rubrica (interattivo)

```python
rubrica = {"Anna": "333-111-2222", "Marco": "347-555-6666", "Sofia": "389-777-8888"}
nome_cercato = input("Cerca contatto: ")
if nome_cercato in rubrica:
    print(f"Numero di {nome_cercato}: {rubrica[nome_cercato]}")
else:
    print(f"{nome_cercato} non è nella rubrica.")
```
