### 13.3 Stampare testo (virgolette doppie o singole)

````python
print("Questo è un testo con virgolette doppie")
print('Questo è un testo con virgolette singole')
````

### 13.4 Stampare numeri (senza virgolette)

````python
print(42)
print(3.14)
print(-100)
````

### 13.5 Stampare più cose insieme (separate da virgola → spazio automatico)

````python
print("Il mio nome è", "Anna")
print("Anni:", 30)
print("Pi greco è circa", 3.14)
````

### 13.6 Andare a capo con \n

````python
print("Prima riga\nSeconda riga\nTerza riga")
````

### 13.7 Riga vuota

````python
print("Sezione 1")
print()
print("Sezione 2")
````
