### 29.2 Metodo classico con for

```python
numeri = [1, 2, 3, 4, 5]
quadrati = []
for numero in numeri:
    quadrati.append(numero ** 2)
print(quadrati)
```

### 29.3 La stessa cosa con una list comprehension

```python
numeri = [1, 2, 3, 4, 5]
quadrati = [numero ** 2 for numero in numeri]
print(quadrati)
```

### 29.4 Con condizione (solo i pari)

```python
numeri = [1, 2, 3, 4, 5, 6]
pari = [numero for numero in numeri if numero % 2 == 0]
print(pari)
```

### 29.5 Dict comprehension

```python
nomi = ["Anna", "Luca", "Marco"]
lunghezze = {nome: len(nome) for nome in nomi}
print(lunghezze)
```

### ESEMPIO PRATICO – prezzi con IVA

```python
prezzi = [10, 20, 30]
prezzi_con_iva = [prezzo * 1.22 for prezzo in prezzi]
print(prezzi_con_iva)
```
