### 18.2 Operazioni di base

```python
a = 10
b = 3
print("Addizione:", a + b)
print("Sottrazione:", a - b)
print("Moltiplicazione:", a * b)
print("Divisione:", a / b)
print("Potenza:", a ** b)
print("Modulo (resto):", a % b)
print("Divisione intera:", a // b)
```

### 18.3 / restituisce sempre float; // dà la parte intera

```python
print(10 / 2)   # 5.0
print(10 // 2)  # 5
print(10 // 3)  # 3
```

### 18.4 Il modulo % (pari/dispari)

```python
print(8 % 2)   # 0 -> pari
print(7 % 2)   # 1 -> dispari
```

### 18.5 Potenza

```python
print(2 ** 10)
print(9 ** 0.5)   # radice quadrata
print(2 ** -1)
```

### 18.6 Priorità (PEMDAS)

```python
print(2 + 3 * 4)      # 14
print((2 + 3) * 4)    # 20
print(2 ** 3 + 1)     # 9
print(10 - 4 / 2)     # 8.0
```

### 18.7 Operatori abbreviati

```python
prezzo = 100
prezzo += 20
print(prezzo)
prezzo -= 10
print(prezzo)
prezzo *= 1.1
print(prezzo)   # 121.00000000000001 (piccola imprecisione dei float)
```

### 18.8 round()

```python
risultato = 10 / 3
print(risultato)
print(round(risultato, 2))
print(round(risultato, 0))
```

### 18.9 Conversione int/float (int tronca, non arrotonda)

```python
print(float(5), type(float(5)))
print(int(3.9), type(int(3.9)))   # 3, non 4
```

### ESEMPIO PRATICO – costo totale di un acquisto

```python
prezzo_unitario = 15.50
quantita = 4
sconto = 0.10

subtotale = prezzo_unitario * quantita
importo_sconto = subtotale * sconto
totale = subtotale - importo_sconto
totale_con_iva = totale * 1.22

print(f"Subtotale: {subtotale:.2f} €")
print(f"Sconto (10%): -{importo_sconto:.2f} €")
print(f"Totale scontato: {totale:.2f} €")
print(f"Totale con IVA 22%: {totale_con_iva:.2f} €")
```
