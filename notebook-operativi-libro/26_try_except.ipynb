### 26.2–26.3 Continuare a chiedere finché l'input è valido (interattivo)

```python
numero_valido = False
while not numero_valido:
    try:
        numero = int(input("Inserisci un numero intero: "))
        numero_valido = True
    except ValueError:
        print("Errore: inserisci un numero intero.")
print("Numero inserito:", numero)
```

### 26.4 else e finally (interattivo)

```python
try:
    numero = int(input("Inserisci un numero: "))
except ValueError:
    print("Valore non valido.")
else:
    print("Conversione riuscita.")
finally:
    print("Fine del controllo.")
```
