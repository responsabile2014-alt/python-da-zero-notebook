### 17.2 Virgolette singole vs doppie

```python
frase = "L'amico di Anna è arrivato"   # apostrofo dentro: uso le doppie
print(frase)
frase2 = 'Mario ha detto "buongiorno" a tutti'
print(frase2)
```

### 17.3 Concatenazione con +

```python
nome = "Anna"
cognome = "Rossi"
nome_completo = nome + " " + cognome
print(nome_completo)
```

Per unire testo e numero serve `str()` (oppure, meglio, le f-string):

```python
eta = 30
messaggio = "Ho " + str(eta) + " anni"
print(messaggio)
```

### 17.4 F-string (il modo moderno)

```python
nome = "Anna"
eta = 30
citta = "Roma"
print(f"Mi chiamo {nome}, ho {eta} anni e abito a {citta}.")
```

```python
prezzo = 19.99
iva = 0.22
print(f"Prezzo: {prezzo} euro")
print(f"IVA (22%): {prezzo * iva:.2f} euro")
print(f"Totale con IVA: {prezzo * (1 + iva):.2f} euro")
```

### 17.5 \n e \t

```python
print("Via Roma, 15\nMilano\nItalia")
```

```python
print("Nome:\tAnna")
print("Età:\t30")
print("Città:\tRoma")
```

### 17.6 Metodi delle stringhe

```python
nome = "anna"
print(nome.upper())
print("ANNA ROSSI".lower())
print("   ciao   ".strip())
print("Ciao, mi chiamo Marco".replace("Marco", "Anna"))
print("mela banana arancia".split())
print("15/03/1994".split("/"))
print("anna@gmail.com".endswith(".com"))
```
