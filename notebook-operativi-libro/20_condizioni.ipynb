### 20.2 if e indentazione (4 spazi)

````python
eta = 20
if eta >= 18:
    print("Sei maggiorenne.")
    print("Puoi votare.")
````

### 20.5 else

````python
eta = 15
if eta >= 18:
    print("Sei maggiorenne.")
else:
    print("Sei minorenne.")
````

### 20.6 elif

````python
voto = 75
if voto >= 90:
    print("Ottimo!")
elif voto >= 75:
    print("Buono.")
elif voto >= 60:
    print("Sufficiente.")
else:
    print("Insufficiente.")
````

### 20.7 Operatori di confronto

````python
a = 10
b = 5
print(a == b)
print(a != b)
print(a > b)
print(a >= 10)
````

### 20.8 Operatori logici and / or / not

````python
eta = 25
reddito = 30000
if eta >= 18 and reddito >= 20000:
    print("Prestito approvato")
else:
    print("Prestito non approvato")
````

### 20.9 Esempio con sconto a soglie

````python
totale = 150.0
if totale >= 200:
    sconto = 0.20
elif totale >= 100:
    sconto = 0.10
else:
    sconto = 0
totale_finale = totale * (1 - sconto)
print(f"Sconto: {int(sconto*100)}%")
print(f"Totale da pagare: {totale_finale:.2f} €")
````

**Errore comune (NON eseguire):** usare `=` invece di `==` in una condizione → `SyntaxError`.
