### 22.2 Accesso per indice (si parte da 0) e indici negativi

```python
frutti = ["mela", "banana", "arancia", "kiwi"]
print(frutti[0])
print(frutti[3])
print(frutti[-1])
print(frutti[-2])
```

### 22.4–22.6 Modificare, aggiungere, rimuovere

```python
frutti = ["mela", "banana", "arancia"]
frutti[1] = "fragola"
print(frutti)
frutti.append("kiwi")
print(frutti)
frutti.insert(1, "pera")
print(frutti)
frutti.remove("fragola")
print(frutti)
rimosso = frutti.pop(1)
print("Rimosso:", rimosso, "->", frutti)
```

### 22.7–22.8 len() e operatore in

```python
frutti = ["mela", "banana", "arancia"]
print(len(frutti))
print("banana" in frutti)
print("kiwi" in frutti)
```

### 22.9 Ciclare con for ed enumerate()

```python
frutti = ["mela", "banana", "arancia"]
for indice, frutto in enumerate(frutti):
    print(f"{indice + 1}. {frutto}")
```

### ESEMPIO PRATICO – lista della spesa

```python
spesa = []
for prodotto in ["pane", "latte", "uova", "pasta", "pomodori"]:
    spesa.append(prodotto)

print("Lista della spesa:")
for i, prodotto in enumerate(spesa, 1):
    print(f"  {i}. {prodotto}")
print(f"Totale prodotti: {len(spesa)}")

spesa.remove("latte")
print("Dopo aver comprato il latte:", spesa)
```
