### 16.2 La funzione type()

```python
print(type("Anna"))
print(type(30))
print(type(19.99))
print(type(True))
```

### 16.3–16.6 str, int, float, bool

```python
nome = "Anna Rossi"
eta = 30
prezzo = 19.99
attivo = True
print(nome, type(nome))
print(eta, type(eta))
print(prezzo, type(prezzo))
print(attivo, type(attivo))
```

### 16.7–16.8 list e tuple

```python
frutta = ["mela", "banana", "arancia"]
print(frutta, type(frutta))

coordinate = (45.4642, 9.1900)  # lat/lon di Milano
print(coordinate, type(coordinate))  # nota: 9.1900 viene mostrato come 9.19
```

### 16.9–16.10 dict e set

```python
persona = {"nome": "Anna", "eta": 30, "citta": "Roma"}
print(persona, type(persona))

colori = {"rosso", "blu", "verde", "rosso"}  # il duplicato sparisce
print(sorted(colori), type(colori))  # sorted() solo per un output prevedibile
```

### 16.11 None

```python
risultato = None
print(risultato, type(risultato))
```

### 16.13 Curiosità: bool è una specializzazione di int

```python
print(True + True)            # 2
print(sum([True, False, True]))  # 2
```
